```python
# ==============================================================================
# IMPORTS & SETUP
# ==============================================================================
import trino
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import ipywidgets as widgets
from IPython.display import display, clear_output
import warnings
warnings.filterwarnings('ignore')

# Thiết lập template màu chuẩn cho AhaMove (Cam, Xanh dương, v.v.)
AHAMOVE_COLORS = ['#f37021', '#00529c', '#4caf50', '#f44336', '#ff9800']

# ==============================================================================
# 1. TRINO DATA PIPELINE (ELT)
# ==============================================================================
def fetch_and_prep_data():
    """
    Kết nối Trino, thực hiện Join fact_orders, fact_stops và dim_date.
    Tạo sẵn các Derived Metrics và Buckets ngay trong SQL để giảm tải cho Pandas.
    """
    # Thay đổi thông tin connection cho phù hợp với môi trường của bạn
    conn = trino.dbapi.connect(
        host="trino-coordinator",
        port=8080,
        user="admin",
        catalog="lakehouse", 
        schema="gold"        
    )
    
    query = """
    WITH stop_metrics AS (
        SELECT 
            order_sk,
            SUM(cod_amount) AS total_cod_amount,
            MAX(stop_index) AS total_stops
        FROM gold.fact_order_stops
        GROUP BY 1
    ),
    raw_data AS (
        SELECT 
            o.order_id,
            CAST(o.create_time AS DATE) AS create_date,
            EXTRACT(HOUR FROM o.create_time) AS create_hour,
            o.service_id,
            o.city_id,
            o.status,
            o.time_to_accept_seconds,
            o.distance_km,
            o.net_revenue,
            o.from_lat,
            o.from_lng,
            COALESCE(s.total_cod_amount, 0) AS total_cod_amount,
            COALESCE(s.total_stops, 1) AS total_stops
        FROM gold.fact_orders o
        LEFT JOIN stop_metrics s ON o.order_sk = s.order_sk
    )
    SELECT * FROM raw_data
    """
    df = pd.read_sql(query, conn)
    
    # ---------------------------------------------------------
    # FEATURE ENGINEERING & BUCKETING (Pandas)
    # ---------------------------------------------------------
    # 1. Time to accept buckets
    cond_t2a = [
        df['time_to_accept_seconds'].isna() | df['status'].isin(['TIMEOUT', 'CANCELLED']),
        df['time_to_accept_seconds'] <= 15,
        df['time_to_accept_seconds'] <= 30,
        df['time_to_accept_seconds'] <= 60,
        df['time_to_accept_seconds'] > 60
    ]
    choice_t2a = ['5. No Accept / Timeout', '1. < 15s', '2. 15-30s', '3. 30-60s', '4. > 60s']
    df['t2a_bucket'] = np.select(cond_t2a, choice_t2a, default='Unknown')
    
    # 2. Unit Price buckets (Net revenue / KM)
    safe_dist = df['distance_km'].replace(0, 0.1) # Tránh lỗi chia 0
    df['unit_price_per_km'] = df['net_revenue'] / safe_dist
    df['price_bucket'] = pd.cut(
        df['unit_price_per_km'],
        bins=[-np.inf, 5000, 6000, 8000, np.inf],
        labels=['< 5k/km', '5k-6k/km', '6k-8k/km', '> 8k/km']
    )
    
    # 3. COD & Stops Buckets
    df['cod_bucket'] = pd.cut(
        df['total_cod_amount'],
        bins=[-1, 1, 500000, 2000000, np.inf],
        labels=['0 COD', '< 500k', '500k - 2M', '> 2M']
    )
    df['stop_bucket'] = pd.cut(
        df['total_stops'],
        bins=[0, 1, 2, 5, np.inf],
        labels=['1 stop', '2 stops', '3-5 stops', '> 5 stops']
    )
    
    # 4. Distance Buckets
    df['distance_bucket'] = pd.cut(
        df['distance_km'],
        bins=[-np.inf, 2, 5, 10, np.inf],
        labels=['Cực ngắn < 2km', 'Ngắn 2-5km', 'Vừa 5-10km', 'Xa > 10km']
    )
    
    return df

# Load dữ liệu vào Memory
df_master = fetch_and_prep_data()
```

```python
# ==============================================================================
# 2. INTERACTIVE DASHBOARD (IPYWIDGETS + PLOTLY)
# ==============================================================================

df_master['create_date'] = pd.to_datetime(df_master['create_date'])

# Khởi tạo Filters
cities = ['All'] + sorted(df_master['city_id'].dropna().unique().tolist())
services = ['All'] + sorted(df_master['service_id'].dropna().unique().tolist())

# Bây giờ hàm .dt.strftime sẽ hoạt động trơn tru
dates = ['All'] + sorted(df_master['create_date'].dt.strftime('%Y-%m-%d').unique().tolist(), reverse=True)

# Khởi tạo UI Widgets
w_city = widgets.Dropdown(options=cities, description='📍 City:', value='All', style={'description_width': 'initial'})
w_service = widgets.Dropdown(options=services, description='🛵 Service:', value='All', style={'description_width': 'initial'})
w_date = widgets.Dropdown(options=dates, description='📅 Date:', value='All', style={'description_width': 'initial'})

output_charts = widgets.Output()

def generate_dashboard(*args):
    with output_charts:
        clear_output(wait=True)
        
        # Apply Filters
        df_filtered = df_master.copy()
        if w_city.value != 'All':
            df_filtered = df_filtered[df_filtered['city_id'] == w_city.value]
        if w_service.value != 'All':
            df_filtered = df_filtered[df_filtered['service_id'] == w_service.value]
        if w_date.value != 'All':
            df_filtered = df_filtered[df_filtered['create_date'].astype(str) == w_date.value]
            
        if df_filtered.empty:
            print("⚠️ Không có dữ liệu cho bộ lọc này. Vui lòng chọn lại!")
            return

        # ======================================================================
        # CHART 1: Market Liquidity Distribution (100% Stacked Bar)
        # ======================================================================
        c1_df = df_filtered.groupby(['create_hour', 't2a_bucket']).size().reset_index(name='orders')
        c1_df['%_total'] = c1_df.groupby('create_hour')['orders'].transform(lambda x: x / x.sum() * 100)
        
        # Sắp xếp lại thứ tự bucket để màu sắc logic (Xanh -> Đỏ)
        color_map = {
            '1. < 15s': '#2ca02c',       # Xanh lá (Hoàn hảo)
            '2. 15-30s': '#1f77b4',      # Xanh dương (Bình thường)
            '3. 30-60s': '#ff7f0e',      # Cam (Chậm)
            '4. > 60s': '#d62728',       # Đỏ (Nguy hiểm)
            '5. No Accept / Timeout': '#7f7f7f' # Xám (Thất bại)
        }
        
        fig1 = px.bar(c1_df, x='create_hour', y='%_total', color='t2a_bucket', 
                      color_discrete_map=color_map,
                      title='<b>Chart 1: Market Liquidity Health (Nhiệt kế Thanh khoản)</b>',
                      labels={'create_hour': 'Giờ trong ngày', '%_total': 'Tỷ trọng Đơn hàng (%)'})
        
        # Vẽ Threshold Line 15% (Cảnh báo Churn rate)
        fig1.add_hline(y=15, line_dash="dash", line_color="red", 
                       annotation_text="Ngưỡng Nguy hiểm (15%)", annotation_position="top left")
        fig1.update_layout(barmode='stack', hovermode="x unified", height=450)

        # ======================================================================
        # CHART 2: Price Elasticity of Acceptance (Độ co giãn Giá)
        # ======================================================================
        c2_df = df_filtered.groupby('price_bucket')['time_to_accept_seconds'].mean().reset_index()
        fig2 = px.line(c2_df, x='price_bucket', y='time_to_accept_seconds', markers=True,
                       title='<b>Chart 2: Price Elasticity of Acceptance (Điểm bùng phát giá)</b>',
                       labels={'price_bucket': 'Đơn giá / Km', 'time_to_accept_seconds': 'Thời gian chờ trung bình (giây)'})
        fig2.update_traces(line_color='#00529c', marker=dict(size=12, color='#f37021'), line_width=3)
        fig2.add_hline(y=120, line_dash="dot", line_color="red", annotation_text="Ngưỡng tài xế 'bơ' đơn (> 120s)")
        fig2.update_layout(height=400)

        # ======================================================================
        # CHART 3: Operational Friction Matrix (Heatmap)
        # ======================================================================
        c3_df = df_filtered.pivot_table(
            index='cod_bucket', columns='stop_bucket', 
            values='time_to_accept_seconds', aggfunc='mean'
        ).fillna(0) # Tránh lỗi NaN khi render
        
        fig3 = go.Figure(data=go.Heatmap(
            z=c3_df.values, x=c3_df.columns, y=c3_df.index,
            colorscale='RdYlGn_r', # Đỏ là tệ (thời gian cao), Xanh là tốt
            text=np.round(c3_df.values, 0), texttemplate="%{text}s",
            hovertemplate='COD: %{y}<br>Stops: %{x}<br>Avg Time to Accept: %{z}s<extra></extra>'
        ))
        fig3.update_layout(title='<b>Chart 3: Operational Friction Matrix (Rào cản Vận hành)</b>',
                           xaxis_title='Số điểm dừng (Stops)', yaxis_title='Giá trị thu hộ (COD)', height=450)

        # ======================================================================
        # CHART 4: Distance Preference & Backhaul Fear (Dual-axis)
        # ======================================================================
        c4_df = df_filtered.groupby('distance_bucket').agg(
            avg_t2a=('time_to_accept_seconds', 'mean'), # Dùng T2A làm proxy cho Ignore Rate
            avg_net_rev=('net_revenue', 'mean')
        ).reset_index()
        
        fig4 = make_subplots(specs=[[{"secondary_y": True}]])
        # Bar: Thời gian chờ nhận đơn (Proxy của độ kén đơn)
        fig4.add_trace(go.Bar(x=c4_df['distance_bucket'], y=c4_df['avg_t2a'], 
                              name="Avg Time to Accept (s)", marker_color='#aec7e8'), secondary_y=False)
        # Line: Thu nhập
        fig4.add_trace(go.Scatter(x=c4_df['distance_bucket'], y=c4_df['avg_net_rev'], 
                                  name="Avg Net Revenue (VND)", mode='lines+markers', 
                                  line=dict(color='#d62728', width=3), marker=dict(size=10)), secondary_y=True)
        
        fig4.update_layout(title='<b>Chart 4: Distance Preference & Nỗi sợ "Chạy Rỗng"</b>', height=450)
        fig4.update_yaxes(title_text="Thời gian chờ nhận đơn (s)", secondary_y=False)
        fig4.update_yaxes(title_text="Thu nhập trung bình (VND)", secondary_y=True)

        # ======================================================================
        # CHART 5: Unfilled Orders Spatial Heatmap (Bản đồ Lỗ đen)
        # ======================================================================
        c5_df = df_filtered[
            (df_filtered['status'].isin(['TIMEOUT', 'CANCELLED'])) | 
            (df_filtered['time_to_accept_seconds'] > 180)
        ].dropna(subset=['from_lat', 'from_lng'])
        
        # FIX CỐT LÕI: Lọc bỏ tọa độ rác, chỉ giữ lại tọa độ nằm trong lãnh thổ Việt Nam
        # Vĩ độ (Lat) VN khoảng: 8.0 đến 24.0
        # Kinh độ (Lng) VN khoảng: 102.0 đến 110.0
        c5_df = c5_df[
            (c5_df['from_lat'].between(8.0, 24.0)) & 
            (c5_df['from_lng'].between(102.0, 110.0))
        ]
        
        if not c5_df.empty:
            # Tối ưu UX: Tự động tính toán tâm bản đồ (Center) dựa trên data thực tế
            # Nếu user chọn filter city='HAN', tâm sẽ tự động focus về Hà Nội và ngược lại
            center_lat = c5_df['from_lat'].mean()
            center_lon = c5_df['from_lng'].mean()
            
            fig5 = px.scatter_mapbox(
                c5_df, lat="from_lat", lon="from_lng", color="create_hour", 
                size_max=15, zoom=11, # Tăng zoom lên 11 để nhìn rõ đường phố HN/HCM hơn
                mapbox_style="carto-positron",
                title='<b>Chart 5: Supply Blackzones (Đơn hàng Timeout hoặc T2A > 180s)</b>',
                color_continuous_scale=px.colors.cyclical.IceFire,
                center={"lat": center_lat, "lon": center_lon} # Bắt buộc bản đồ phải focus vào tâm này
            )
            fig5.update_layout(margin={"r":0,"t":40,"l":0,"b":0}, height=500)
        else:
            # Xử lý an toàn khi UI không có đơn lỗi nào ở khu vực đó
            fig5 = go.Figure()
            fig5.update_layout(
                title='<b>Chart 5: Supply Blackzones</b><br>Khu vực này hiện không có "lỗ đen" nguồn cung nào.', 
                height=200,
                xaxis=dict(visible=False), 
                yaxis=dict(visible=False)
            )
        # ======================================================================
        # RENDER DASHBOARD
        # ======================================================================
        fig1.show()
        fig2.show()
        fig3.show()
        fig4.show()
        fig5.show()

# Binding Events
w_city.observe(generate_dashboard, names='value')
w_service.observe(generate_dashboard, names='value')
w_date.observe(generate_dashboard, names='value')

# Layout UI
ui_filters = widgets.HBox([w_date, w_city, w_service], layout=widgets.Layout(margin='0 0 20px 0'))
display(widgets.VBox([ui_filters, output_charts]))

# Init call
generate_dashboard()
```
